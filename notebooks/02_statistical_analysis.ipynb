# 02 Statistical Analysis
This notebook computes descriptive statistics and generates visualizations for the selected 100 startup projects.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

df = pd.read_csv("../data/processed/startup_projects.csv")
print("Dataset Shape:", df.shape)
df.head()
```

## Core Numerical Statistics

```python
df[[
    "investment_cost",
    "funding_rounds",
    "funding_duration_years"
]].describe()
```

## Investment Statistics

```python
print("Total investment:", df["investment_cost"].sum())
print("Average investment:", df["investment_cost"].mean())
print("Median investment:", df["investment_cost"].median())
print("Minimum investment:", df["investment_cost"].min())
print("Maximum investment:", df["investment_cost"].max())
```

## Sector Statistics

```python
sector_stats = df.groupby("sector").agg(
    projects=("project_id", "count"),
    total_investment=("investment_cost", "sum"),
    average_investment=("investment_cost", "mean"),
    median_investment=("investment_cost", "median"),
    average_rounds=("funding_rounds", "mean")
)
sector_stats
```

## Funding Band Statistics

```python
band_stats = df.groupby("funding_band").agg(
    projects=("project_id", "count"),
    total_investment=("investment_cost", "sum"),
    average_investment=("investment_cost", "mean"),
    median_investment=("investment_cost", "median")
)
band_stats
```

## Acquisition and IPO Statistics

```python
print("Acquired companies:", df["acquisition"].sum())
print("IPO companies:", df["ipo"].sum())
print("\nSummary of outcomes:")
print(df[["acquisition", "ipo"]].sum())
```

## Chart 1: Projects by Sector

```python
sector_counts = df["sector"].value_counts()

plt.figure(figsize=(10, 6))
sector_counts.plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Number of Selected Investment Projects by Sector")
plt.xlabel("Sector")
plt.ylabel("Number of Projects")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Chart 2: Total Investment by Sector

```python
sector_investment = (
    df.groupby("sector")["investment_cost"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))
sector_investment.plot(kind="bar", color="salmon", edgecolor="black")
plt.title("Total Historical Funding by Sector")
plt.xlabel("Sector")
plt.ylabel("Funding (USD)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Chart 3: Investment Cost Distribution

```python
plt.figure(figsize=(10, 6))
plt.hist(df["investment_cost"], bins=20, color="lightgreen", edgecolor="black")
plt.xscale("log")
plt.title("Distribution of Investment Cost")
plt.xlabel("Investment Cost (USD, log scale)")
plt.ylabel("Number of Projects")
plt.tight_layout()
plt.show()
```

## Chart 4: Funding Rounds vs Investment Cost

```python
plt.figure(figsize=(10, 6))
plt.scatter(
    df["funding_rounds"],
    df["investment_cost"],
    color="purple",
    alpha=0.7,
    edgecolors="black"
)

plt.yscale("log")
plt.title("Funding Rounds vs Investment Cost")
plt.xlabel("Number of Funding Rounds")
plt.ylabel("Investment Cost (USD, log scale)")
plt.tight_layout()
plt.show()
```

## Chart 5: Observed Startup Outcomes (Acquisitions & IPOs)

```python
outcomes = pd.Series({
    "Acquired": df["acquisition"].sum(),
    "IPO": df["ipo"].sum()
})

plt.figure(figsize=(6, 5))
outcomes.plot(kind="bar", color=["gold", "lightcoral"], edgecolor="black")
plt.title("Observed Startup Outcomes")
plt.ylabel("Number of Projects")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
